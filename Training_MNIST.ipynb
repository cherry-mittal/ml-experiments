```python
import torch
import torch.nn as nn
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
```

```python
#getting the dataset and storing it in our memory
train_dataset = datasets.MNIST(root = "./data",
                               train = True,
                               download = True,
                               transform=transforms.ToTensor())
test_dataset = datasets.MNIST(root= "./data",
                              train = False,
                              download = True,
                              transform=transforms.ToTensor())
```

```python
print(train_dataset)
print(test_dataset)
```

```python
#loading the dataset from memory
train_load = DataLoader(train_dataset, batch_size = 64, shuffle = True)
test_load = DataLoader(test_dataset)
print(len(train_load))
```

```python
#defining MLP class
class MNISTClass(nn.Module):
  def __init__(self):
    super().__init__()
    self.linear1 = nn.Linear(in_features=784, out_features=256)
    self.relu = nn.ReLU()
    self.linear2 = nn.Linear(256, 10)

  def forward(self, x):
    x = x.view(x.size(0), -1)
    x = self.linear1(x)
    x = self.relu(x)
    x = self.linear2(x)
    return x
```

```python
model = MNISTClass()
lossfn = nn.CrossEntropyLoss()
optimizer = torch.optim.AdamW(model.parameters(), lr=0.001, weight_decay=0.01)

for epoch in range(20):
  for train, label in train_load:
    pred = model(train)
    loss = lossfn(pred, label)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

  print(f"epoch: {epoch}  loss: {loss.item()}")
```

```python
correct_val = 0
total = 0
with torch.no_grad():
  for test, label in test_load:
    pred = model(test)
    pred = torch.argmax(pred)
    correct_val += (pred == label).item()
    total += 1

print(f"correct values: {correct_val}")
print(f"totoal values: {total}")
```
